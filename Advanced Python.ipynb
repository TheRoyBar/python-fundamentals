```python
#Generators - simpler way to use Iterators in python. Using Lazy Loading
```

```python
def square(n):
    for i in range (5):
        yield i**2
```

```python
gen = square(4)
```

```python
#next(gen)
```

```python
#We can also yield multiple values:
def my_gen(n):
    yield 1
    yield 2
    
```

```python
gen = my_gen(4)
```

```python
next(gen)
```

```python
next(gen)
```
