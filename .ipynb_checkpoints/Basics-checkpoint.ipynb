```python
#Use a dictionary to count the frequency of elements in a list
numbers = [1,2,3,2,3,3,4,5,6,6,6,6,6,4,5]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number] += 1;
    else:
        frequency[number] = 1;

print(frequency)
```

```python
frequency
```

```python
fre2 = {7:1,9:0,8:0}
```

```python
fre2
```

```python
print({**frequency,**fre2}) #Keyvalue arguments, i.e, the argument is in key:value pairs.
```
