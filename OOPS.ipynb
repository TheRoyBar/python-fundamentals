Inheritance

```python
class Car:
    def __init__(self, name, type, enginetype):
        self.name = name
        self.type = type
        self.enginetype= enginetype

    def owner(self):
        print(f"The owner of the car is {self.name}.")
```

```python
audi = Car("Roy", "Petrol", "v8")
```

```python
class Tesla(Car):
    def __init__(self, name, type, enginetype, is_selfDriving):
        super().__init__(name, type, enginetype)
        self.is_selfDriving = is_selfDriving

    def self_driving(self):
        print("The car can self drive.")
```

```python
ModelY = Tesla("Shubham", "Electric", "Induction Motored", 'True')
```

```python
ModelY.self_driving()
```

```python
class Animal():
    def __init__(self, species, leg_numbers):
        self.species = species
        self.leg_numbers = leg_numbers

    def is_species(self):
        print(f'The animal is a {self.species}.')
```

```python
class Pet():
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def barks(self):
        print(f'The name is: {self.name}')
```

```python
class Dog(Animal, Pet):
    def __init__(self, name, age, species, family_address):
        Animal.__init__(self, species, leg_numbers=4)
        Pet.__init__(self, name, age)

    def introduce(self):
        print(f"The name is: {self.name} and my age is {self.age}.")
```

Using Method Overloading

```python
class College():
    def syllabus(self):
        pass

class Makaut(College):
    def syllabus(self):
        return "Makaut University Syllabus"

class IITG(College):
    def syllabus(self):
        return "IIT Syllabus"
```

```python
Student2.syllabus()
```

Using Interfaces/ Abstract Base Classes

```python
from abc import ABC, abstractmethod
```

```python
class Shape(ABC):
    @abstractmethod

    def area(self):
        return "Calculates Area"

class Rectangle(Shape):
    def area(self, h, w):
        self.h = h
        self.w = w
        return "Area of Circle: ",h*w

class Circle(Shape):
    def area(self, r):
        self.r = r
        a = 3.14*self.r*self.r
        return "Area of Circle: ",a
```

```python
c1 = Circle()
r1 = Rectangle()
```

```python
c1.area(4)
```

```python
r1.area(3,5)
```

Although it is supposed to be a private variable we can still access it from its edited name

```python
# Defining the class

class Pan():
    def __init__(self, name, age, panNumber):
        self.name = name # Public Access Modifier
        self._age = age # Protected Access Modifier
        self.__panNumber = panNumber # Private Access Modifier
        
    def get_pan(self):
        return self.__panNumber

    def set_pan(self, new):
        self.__panNumber = new
```

```python
person1 = Pan("Roy", 20, 'YPH322B45')
```

```python
person1.get_pan()
```

```python
person1.set_pan("YPH322B49")
```

```python
person1.get_pan()
```

Magic Methods (also called as Dunder Methods)

```python
class Person():
    def __init__(self,name,age,gender):
        self.name = name
        self.age = age
        self.gender = gender

    def __str__(self):
        return f"The person's name is {self.name}"

    def __repr__(self):
        return (f"I am {self.name}, {self.age} years old.")
```

```python
person = Person("Roy",20,"Male")
```

```python
print(person)
```
